# **Salary Prediction System Based on Years using Simple Linear Regression**

**Step 1: Importing Libraries**

```python
import numpy as np  ##Alice name
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv("Salary_Data.csv")
```

```python
data.shape
```

```python
data
```

**Step 2: Input/Output Split**

*   Input / Independent / X
*   Output / Dependent / Y



```python
Independent = data[["YearsExperience"]]
```

```python
Independent
```

```python
Dependent = data[["Salary"]]
```

```python
Dependent
```

**Step 3: Train Test Split**

```python
#### Sci-Kit Learn
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_Train, X_Test, y_Train, y_Test = train_test_split(Independent, Dependent, test_size=0.40, random_state=0)
```

```python
X_Train.shape
```

```python
X_Test.shape
```

```python
y_Train.shape
```

```python
y_Test.shape
```

**Step 4: Algorithm**

```python
from sklearn.linear_model import LinearRegression
```

```python
regressor = LinearRegression()
```

```python
regressor.fit(X_Train,y_Train)
```

**Step 5: To get the Weight and Bias Value**

```python
weight = regressor.coef_
```

```python
weight
```

```python
bias = regressor.intercept_
```

```python
bias
```

```python
y_pred = regressor.predict(X_Test)
```

**Step 6: To Find the Evaluation metrics**

```python
from sklearn.metrics import r2_score
```

```python
r_score = r2_score(y_Test,y_pred)
r_score
```

**Step 7: To Save the Model**

```python
import pickle
```

```python
filename = "SLR-model.pkl"
```

```python
pickle.dump(regressor,open(filename,'wb'))
```

```python
loaded_model = pickle.load(open("SLR-model.pkl",'rb'))
```

```python

Year = int(input("Enter the Year:"))
result = loaded_model.predict([[Year]])
print(f"The Overall Predicted Salary for {Year} is",result)
```
